```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

# Llama3.2-1B-IT

```python
results_llama3_2_1b_it = pd.read_csv('/content/results_PhoMT_llama3.2-1b-it.csv')
```

```python
# Get descriptive statistics for the COMET-22 score
comet_stats = results_llama3_2_1b_it['COMET_22'].describe()
print("COMET-22 Score Statistics:")
print(comet_stats)
```

```python
# Visualize the distribution
plt.figure(figsize=(10, 6))
sns.histplot(results_llama3_2_1b_it['COMET_22'], kde=True, color='skyblue')
plt.title('Distribution of COMET-22 Scores (Llama3.2-1B-IT)')
plt.xlabel('COMET-22 Score')
plt.ylabel('Frequency')
plt.grid(axis='y', linestyle='--', alpha=0.7)
plt.show()
```

```python
plt.figure(figsize=(8, 5))
sns.boxplot(x=results_llama3_2_1b_it['COMET_22'], color='lightgreen')
plt.title('Boxplot of COMET-22 Scores (Llama3.2-1B-IT)')
plt.xlabel('COMET-22 Score')
plt.grid(axis='x', linestyle='--', alpha=0.7)
plt.show()
```
